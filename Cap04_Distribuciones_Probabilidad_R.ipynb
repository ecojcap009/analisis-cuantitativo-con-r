# 3. Distribuciones de probabilidad discretas
## 3.1. Permutaciones, combinaciones y el cociente binomial (pág. 154-155)

### Conceptos clave
Antes de estudiar las distribuciones discretas es necesario comprender:
- **Permutaciones:** ordenaciones de _k_ elementos tomados de _n_, donde el orden importa.
  Fórmula: P(n, k) = n! / (n − k)!
- **Combinaciones:** subconjuntos de _k_ elementos tomados de _n_, donde el orden **no** importa.
  La función `combn(x, k)` devuelve una matriz con todas las combinaciones posibles.
- **Cociente binomial** C(n, k) = n! / [k! · (n−k)!]:
  número de combinaciones de _n_ elementos tomados de _k_ en _k_. En R: `choose(n, k)`.

### Dependencias
Todas las funciones pertenecen al paquete `base` (no requieren instalación adicional).

```r
# ---------------------------------------------------------------
# 3.1a Funcion de permutaciones (pag. 154)
# P(n, k) = n! / (n - k)!
# Parametros:
#   n : numero total de elementos disponibles
#   k : numero de posiciones a llenar
# Valor de retorno: numero entero (permutaciones posibles)
# ---------------------------------------------------------------

# Definicion de la funcion perm
perm <- function(n, k) factorial(n) / factorial(n - k)

# Ejemplo: 7 personas, 3 sillas -> 7!/4! = 210 permutaciones
perm(7, 3)
# [1] 210
```

```r
# ---------------------------------------------------------------
# 3.1b Combinaciones con combn() (pag. 155)
# combn(x, k): devuelve una MATRIZ donde cada columna es
# una combinacion distinta de k elementos tomados del vector x.
# El orden no importa: {a,b} = {b,a}.
# ---------------------------------------------------------------

# Crear vector de personas usando las 3 primeras letras del abecedario
# letters es un vector built-in de R con las 26 letras minusculas
(personas <- letters[1:3])
# [1] "a" "b" "c"

# Numero total de personas
n <- length(personas)

# Numero de posiciones a asignar
k <- 2

# Todas las combinaciones de 2 elementos sobre 3 personas
# Resultado: matriz 2x3 -- cada columna es una combinacion
combn(personas, k)
#      [,1] [,2] [,3]
# [1,] "a"  "a"  "b"
# [2,] "b"  "c"  "c"
```

```r
# ---------------------------------------------------------------
# 3.1c Cociente binomial con choose() (pag. 155)
# choose(n, k) = C(n, k) = n! / [k! * (n-k)!]
# Devuelve el numero TOTAL de combinaciones posibles.
# Es equivalente a ncol(combn(n, k)).
# ---------------------------------------------------------------

# C(3, 2) = 3 combinaciones posibles de 2 elementos sobre 3
choose(3, 2)
# [1] 3
```

## 3.2. Distribución Binomial (pág. 156-159)

### Conceptos clave
Una variable aleatoria X ~ B(n, p) cuenta el número de éxitos en _n_ ensayos de Bernoulli
independientes, cada uno con probabilidad de éxito _p_.

- **Media:** E(X) = n·p
- **Varianza:** V(X) = n·p·(1−p)
- **Función de cuantía:** P(X = x) = C(n,x) · p^x · (1−p)^(n−x)

Funciones R: `dbinom()`, `pbinom()`, `qbinom()`, `rbinom()`

### Advertencia
> ⚠️ `lower.tail=TRUE` (default) calcula P(X ≤ x). Con `lower.tail=FALSE` se obtiene P(X > x).

```r
# ---------------------------------------------------------------
# 3.2a Funcion de cuantia binomial: dbinom() (pag. 157)
# Ejemplo: jugador de baloncesto, p=0.3, n=5 lanzamientos.
# Calcula P(X = x) para x = 0, 1, 2, 3, 4, 5.
# ---------------------------------------------------------------

# Calcular probabilidades para cada numero de triples anotados (0 a 5)
# dbinom(vector_x, size=n_ensayos, prob=p_exito)
ensayo <- dbinom(0:5, size = 5, prob = 0.3)

# Asignar nombres a las posiciones del vector resultado
rownames(ensayo) <- 0:5   # asignacion alternativa vista en el libro
names(ensayo)    <- 0:5   # forma canonica para vectores

# Mostrar el vector con probabilidades
print(ensayo)
#         0         1         2         3         4         5
# 0.16807 0.36015 0.30870 0.13230 0.02835 0.00243
```

```r
# ---------------------------------------------------------------
# 3.2b Probabilidad acumulada cola izquierda: pbinom() (pag. 157-158)
# lower.tail=TRUE (default): calcula P(X <= x)
# ---------------------------------------------------------------

# P(X <= x) para x = 0, 1, 2, 3, 4, 5
pbinom(0:5, size = 5, prob = 0.3, lower.tail = TRUE)
# [1] 0.16807 0.52822 0.83692 0.96922 0.99757 1.00000
```

```r
# ---------------------------------------------------------------
# 3.2c Probabilidad acumulada cola derecha: pbinom() (pag. 158)
# lower.tail=FALSE: calcula P(X > x)
# Nota: P(X <= x) + P(X > x) = 1 siempre.
# ---------------------------------------------------------------

# P(X > x) para x = 0, 1, 2, 3, 4, 5
pbinom(0:5, size = 5, prob = 0.3, lower.tail = FALSE)
# [1] 0.83193 0.47178 0.16308 0.03078 0.00243 0.00000
```

```r
# ---------------------------------------------------------------
# 3.2d Grafico de la probabilidad acumulada cola derecha (pag. 158)
# Parametros graficos:
#   type="h" : lineas verticales (histograma)
#   lwd=4    : grosor de linea
# ---------------------------------------------------------------

plot(pbinom(0:5, size = 5, prob = 0.3,
     lower.tail = FALSE), type = "h",
     lwd = 4, xlab = "Numero de exitos",
     ylab = "Probabilidad",
     main = "Probabilidad acumulada (cola derecha)")
```

## 3.3. Distribución Geométrica (pág. 159-160)

### Conceptos clave
La distribución geométrica modela el **número de ensayos necesarios** hasta obtener el primer éxito.
- P(X = x) = (1 − p)^(x−1) · p
- P(X ≤ x) = 1 − (1 − p)^x

### Advertencia importante sobre `pgeom()`
> ⚠️ En R, el primer argumento de `pgeom()` es el número de **fallos** antes del primer éxito
> (es decir, x − 1, no x). En el ejemplo del libro: P(Y ≤ 6) requiere introducir 6 − 1 = 5.

```r
# ---------------------------------------------------------------
# 3.3 Distribucion Geometrica: pgeom() (pag. 160)
# Ejemplo: mismo jugador de baloncesto, p=0.3.
# X = numero del intento en que encesta el PRIMER triple.
# Queremos P(Y <= 6) = F(6) = 1 - (1 - 0.3)^6 = 0.8823
# En pgeom(): primer arg = n de FALLOS antes del 1er exito = 6-1 = 5
# ---------------------------------------------------------------

# P(Y <= 6): probabilidad de encestar el primer triple
# dentro de los primeros 6 lanzamientos
pgeom(5, prob = 0.3, lower.tail = TRUE)
# [1] 0.882351
```

## 3.4. Distribución Hipergeométrica (pág. 160-162)

### Conceptos clave
Modela el número de éxitos en una extracción **sin reposición** de _n_ elementos de una
población de N elementos (N₁ del tipo buscado y N₂ = N − N₁ del otro tipo).

Función R: `dhyper(x, m, n, k)` y `phyper(x, m, n, k, lower.tail)` donde:
- `m` = N₁ (elementos del tipo «éxito» en la población)
- `n` = N₂ (elementos del tipo «fracaso» en la población)
- `k` = tamaño de la extracción

### Advertencia
> ⚠️ Con `lower.tail=FALSE` R calcula P(X > x). Para obtener P(X ≥ x) en variables
> discretas hay que introducir x − 1. Ej.: P(X ≥ 2) = P(X > 1) → introducir x = 1.

```r
# ---------------------------------------------------------------
# 3.4a Funcion de cuantia hipergeometrica: dhyper() (pag. 161)
# Ejemplo: opositor que ha preparado 12 de 18 temas.
# El examen extrae 3 temas al azar SIN reposicion.
#   m = N1 = 12 (temas preparados)
#   n = N2 = 6  (temas no preparados)
#   k = 3       (temas que entran en el examen)
# ---------------------------------------------------------------

# Calcular P(X = x) para x = 0, 1, 2, 3
ensayo <- dhyper(0:3, m = 12, n = 6, k = 3)

# Asignar nombres a las posiciones del vector
names(ensayo) <- 0:3

# Mostrar las probabilidades
print(ensayo)
#      0      1      2      3
# 0.0245 0.2205 0.4853 0.2696
```

```r
# ---------------------------------------------------------------
# 3.4b Probabilidad acumulada cola derecha: phyper() (pag. 162)
# P(X >= 2) = P(X > 1) con lower.tail=FALSE
# Al trabajar con distribuciones DISCRETAS hay que tener cuidado:
# lower.tail=FALSE calcula P(X > x), NO P(X >= x).
# Por eso se introduce x=1 para obtener P(X >= 2).
# ---------------------------------------------------------------

# Probabilidad de conocer al menos 2 de los 3 temas del examen
phyper(1, m = 12, n = 6, k = 3, lower.tail = FALSE)
# [1] 0.754902
```

## 3.5. Distribución de Poisson (pág. 162-163)

### Conceptos clave
La distribución de Poisson modela el número de eventos en un intervalo de tiempo fijo,
dado que los eventos ocurren con una tasa media constante λ.

- **Único parámetro:** λ > 0
- **Media y varianza coinciden:** E(X) = V(X) = λ
- **Función de cuantía:** P(X = k) = (λ^k · e^−λ) / k!

Funciones R: `dpois(x, lambda)`, `ppois(x, lambda, lower.tail)`

```r
# ---------------------------------------------------------------
# 3.5a Funcion de cuantia Poisson: dpois() (pag. 163)
# Ejemplo: tiempo de espera del autobus ~ Poisson(lambda=3)
# Calcula P(X = 5): probabilidad de esperar exactamente 5 minutos.
# Formula: P(X=5) = (3^5 * e^(-3)) / 5! = 0.1008
# ---------------------------------------------------------------

# Probabilidad puntual en x=5 con lambda=3
dpois(5, 3)
# [1] 0.1008188
```

```r
# ---------------------------------------------------------------
# 3.5b Grafico de la funcion de densidad Poisson (pag. 163)
# Representa la funcion de cuantia para x en [0, 12] con lambda=3.
# ---------------------------------------------------------------

# Vector de valores de x: enteros de 0 a 12
x <- 0:12

# Calcular la funcion de cuantia evaluada en cada valor de x
ensayo <- dpois(0:12, 3)

# Representacion grafica: type="o" = puntos unidos por lineas
plot(x, ensayo, type = "o",
     xlab = "Valor de x", ylab = "Probabilidad",
     lwd = 3)
```

# 4. Distribuciones de probabilidad continuas
## 4.1. Distribución Uniforme Continua (pág. 164-166)

### Conceptos clave
X ~ U(a, b): todos los intervalos de igual longitud dentro del rango [a, b] son igualmente probables.

- **Función de densidad:** f(x) = 1/(b−a) para a ≤ x ≤ b
- **Media:** E(X) = (a + b) / 2
- **Varianza:** V(X) = (b−a)² / 12

Función R: `punif(x, min, max, lower.tail)`

### Advertencia
> ⚠️ En distribuciones **continuas**, P(X < x) = P(X ≤ x) porque la probabilidad
> de un punto exacto es siempre cero. No existe distinción entre cola estricta e inclusiva.

```r
# ---------------------------------------------------------------
# 4.1a Cola izquierda uniforme: punif() (pag. 165)
# Ejemplo: hora del parto X ~ U(0, 24).
# P(X < 5): probabilidad de dar a luz dentro de las primeras 5 horas.
# Resultado esperado: 5/24 ~= 0.2083
# ---------------------------------------------------------------

# P(X < 5) con distribucion Uniforme(0, 24)
punif(5, min = 0, max = 24, lower.tail = TRUE)
# [1] 0.2083333
```

```r
# ---------------------------------------------------------------
# 4.1b Cola derecha uniforme: punif() (pag. 165-166)
# P(X > 22): probabilidad de dar a luz en las ultimas 2 horas.
# Resultado esperado: 2/24 ~= 0.0833
# ---------------------------------------------------------------

# P(X > 22) con distribucion Uniforme(0, 24)
punif(22, min = 0, max = 24, lower.tail = FALSE)
# [1] 0.08333333
```

## 4.2. Distribución Exponencial (pág. 166-168)

### Conceptos clave
Relacionada con la distribución de Poisson: modela el **tiempo entre eventos** sucesivos.

- **Parámetro:** θ = λ > 0 (tasa de ocurrencia)
- **Media:** E(X) = 1/θ
- **Varianza:** V(X) = 1/θ²
- **Función de distribución:** F(x) = P(X ≤ x) = 1 − e^(−θx)

Funciones R: `pexp(x, rate, lower.tail)`, `dexp(x, rate)`

```r
# ---------------------------------------------------------------
# 4.2a Cola derecha exponencial: pexp() (pag. 167)
# Ejemplo: tiempo entre partos ~ Exponencial(theta=4).
# E(X) = 1/4 = 0.25 horas (15 minutos) entre partos.
# P(X > 1): probabilidad de que transcurra mas de 1 hora entre partos.
# ---------------------------------------------------------------

# P(X > 1): cola derecha con rate=theta=4
pexp(1, 4, lower.tail = FALSE)
# [1] 0.01831564
```

```r
# ---------------------------------------------------------------
# 4.2b Cola izquierda exponencial: pexp() (pag. 167)
# P(X < 1) = 1 - P(X > 1): complementaria del calculo anterior.
# ---------------------------------------------------------------

# P(X < 1): probabilidad de que transcurra MENOS de 1 hora entre partos
pexp(1, 4, lower.tail = TRUE)
# [1] 0.9816844
```

```r
# ---------------------------------------------------------------
# 4.2c Grafico de la funcion de densidad exponencial (pag. 167-168)
# Representa dexp(x, 4) para x en [0, 2] con paso 0.1.
# ---------------------------------------------------------------

# Crear secuencia de valores x de 0 a 2 con paso 0.1
x <- seq(0, 2, by = 0.1)

# Graficar la funcion de densidad exponencial con theta=4
plot(x, dexp(x, 4), type = "l", lwd = 3,
     xlab = "Valores de x", ylab = "Densidad",
     main = "Funcion de densidad de la funcion exponencial")
```

## 4.3. Distribución Normal (pág. 168-174)

### Conceptos clave
La distribución más importante de la estadística. X ~ N(μ, σ):

- **Parámetros:** media μ y desviación estándar σ
- **Función de densidad (campana de Gauss):** f(x) = [1/(σ√2π)] · e^[−½((x−μ)/σ)²]
- Es **simétrica** respecto a μ: P(X ≤ μ) = P(X ≥ μ) = 0.5

Funciones R: `dnorm()` (densidad), `pnorm()` (acumulada), `qnorm()` (cuantil)

### Advertencia
> ⚠️ En R, el argumento `sd` de estas funciones siempre es la **desviación estándar** σ,
> nunca la varianza σ².

```r
# ---------------------------------------------------------------
# 4.3a Funcion de densidad normal: dnorm() y plot() (pag. 169-170)
# Ejemplo: peso de bebes al nacer ~ N(mu=3.5, sigma=0.25)
# Grafica la funcion de densidad para x en [2, 5].
# ---------------------------------------------------------------

# Crear vector de valores de x
x <- seq(2, 5, by = 0.1)

# Calcular la funcion de densidad para cada valor de x
densidad <- dnorm(x, 3.5, 0.25)

# Graficar la curva de densidad
plot(x, densidad, type = "l", lwd = 3,
     xlab = "Variable x (peso al nacer)",
     ylab = "Densidad",
     main = "Funcion de densidad normal")

# Aniadir linea vertical discontinua en la media (lty=2)
abline(v = 3.5, lty = 2)
```

```r
# ---------------------------------------------------------------
# 4.3b Probabilidad de un intervalo [3, 4]: pnorm() (pag. 171)
# Calcula P(3 <= X <= 4) usando la simetria de la normal:
#   P(3 <= X <= 4) = 1 - 2 * P(X <= 3)
# ---------------------------------------------------------------

# Paso 1: calcular P(X <= 3) con la cola izquierda
pnorm(3, mean = 3.5, sd = 0.25, lower.tail = TRUE)
# [1] 0.02275013

# Paso 2 (implicito en el texto): P(3 <= X <= 4) = 1 - 2 * P(X <= 3)
# Resultado: 1 - 2 * 0.023 = 0.954 (mas del 95% de los bebes)
```

```r
# ---------------------------------------------------------------
# 4.3c Calculo de cuantiles (percentiles): qnorm() (pag. 172)
# El doctor necesita los percentiles 5% y 20% del peso de los bebes
# para clasificarlos en 3 grupos.
# ---------------------------------------------------------------

# Percentiles 5% (p=0.05) y 20% (p=0.20) de la normal N(3.5, 0.25)
qnorm(c(0.05, 0.2), mean = 3.5, sd = 0.25,
      lower.tail = TRUE)
# [1] 3.088787 3.289595
# Interpretacion:
#   Grupo "muy poco peso" : < 3.09 kg
#   Grupo "poco peso"     : 3.09 - 3.29 kg
#   Grupo "peso normal"   : > 3.29 kg
```

```r
# ---------------------------------------------------------------
# 4.3d Distribucion muestral de la media, sigma conocida (pag. 173)
# X_bar ~ N(mu, sigma/sqrt(n)) cuando sigma es conocida.
# Ejemplo: n=150 bebes, mu=3.5, sigma=0.25
# ---------------------------------------------------------------

# P(X_bar >= 3.75): cola derecha de la distribucion muestral
# sd = error estandar = sigma / sqrt(n) = 0.25 / sqrt(150) ~= 0.02
pnorm(3.75, mean = 3.5, sd = 0.02, lower.tail = FALSE)
# [1] 3.732564e-36   <- practicamente cero (notacion cientifica)
```

## 4.4. Distribución t de Student (pág. 174-175)

### Conceptos clave
Se usa cuando σ es **desconocida** y debe estimarse con la desviación muestral _s_.
Si X ~ N(μ, σ) y σ es desconocida, entonces t = (X̄ − μ)/(s/√n) ~ t_{n-1}

Función R: `pt(t, df, lower.tail)` donde `df` = grados de libertad = n − 1

### Advertencia
> ⚠️ El estadístico _t_ puede ser negativo. En el ejemplo del libro el valor es −34.02,
> lo que produce una probabilidad extremadamente pequeña (≈ 0).

```r
# ---------------------------------------------------------------
# 4.4 t de Student: pt() (pag. 175)
# Continuacion del ejemplo del peso de bebes, pero ahora sigma
# es DESCONOCIDA y se estima con s = 0.18.
#
# Queremos P(X_bar <= 3) con n=150, mu=3.5, s=0.18.
# Transformacion a t:
#   t = (x - mu) / (s / sqrt(n))
#     = (3 - 3.5) / (0.18 / sqrt(150))
#     = -0.5 / 0.01470
#     = -34.02
# ---------------------------------------------------------------

# P(T <= -34.02) con 149 grados de libertad
# df = n - 1 = 150 - 1 = 149
pt(-34.02, df = 149, lower.tail = TRUE)
# [1] 1.973326e-72   <- probabilidad practicamente nula
```

## 4.5. Teorema del Límite Central - TLC (pág. 176-178)

### Conceptos clave
El TLC establece que, para muestras suficientemente grandes (n > 30), la distribución de la
media muestral se aproxima a una **normal**, independientemente de la distribución original:
X̄ ≈ N(μ, σ/√n)

**Aplicación a la Binomial:** cuando n > 30 y np > 5 y n(1−p) > 5:
X ~ B(n, p) ≈ N(np, √[np(1−p)])

**Aplicación a proporciones:** la proporción muestral p̂ ~ N(p, √[p(1−p)/n])

### Advertencia
> ⚠️ El argumento `sd` de `pnorm()` siempre recibe la **desviación estándar** (√varianza),
> no la varianza directamente. En el ejemplo: sd = sqrt(10.5), no 10.5.

```r
# ---------------------------------------------------------------
# 4.5a TLC aplicado a la Binomial (pag. 176)
# Binomial B(50, 0.3) aproximada a Normal.
#   E(X) = n*p = 50*0.3 = 15
#   Var(X) = n*p*(1-p) = 50*0.3*0.7 = 10.5
#   sd = sqrt(10.5)
# Queremos P(X > 20): jugador encesta mas de 20 de 50 triples.
# ---------------------------------------------------------------

# P(X > 20) usando la aproximacion normal
pnorm(c(20), mean = 15, sd = sqrt(10.5),
      lower.tail = FALSE)
# [1] 0.06141132
# Interpretacion: ~6% de probabilidad de encestar mas de 20 triples en 50 intentos
```

```r
# ---------------------------------------------------------------
# 4.5b TLC aplicado a proporciones / Bernoulli (pag. 177-178)
# La proporcion de triples Y = X/n sigue aproximadamente N(p, sqrt(p*(1-p)/n)).
#   E(Y) = p = 0.3
#   Var(Y) = p*(1-p)/n = 0.3*0.7/50 = 0.0042
# Queremos P(Y > 0.4): proporcion de triples superior al 40%.
# El resultado debe coincidir con el de 4.5a: 40% de 50 = 20 triples.
# ---------------------------------------------------------------

# P(Y > 0.4): cola derecha de la proporcion muestral
pnorm(0.4, mean = 0.3, sd = sqrt(0.0042),
      lower.tail = FALSE)
# [1] 0.06141132
# El resultado es identico al anterior: verificacion del TLC.
```
